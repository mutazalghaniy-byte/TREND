```python
import os
import pandas as pd
from IPython.display import display

# 1. Menentukan nama/lokasi file
file_path = "klasifikasimhs.csv"

# 2. Memeriksa ketersediaan file dan membacanya
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("Dataset berhasil dibaca!\n")
    
    # 3. Menjalankan kode ringkasan yang Anda buat sebelumnya
    nama_dataset = file_path
    ringkasan = pd.DataFrame({
        "Informasi": [
            "Nama Dataset",
            "Jumlah Baris",
            "Jumlah Kolom",
            "Jumlah Data Kosong",
            "Jumlah Baris Duplikat",
            "Jumlah Kolom Numerik",
            "Jumlah Kolom Kategorikal"
        ],
        "Hasil": [
            nama_dataset,
            df.shape[0],
            df.shape[1],
            df.isnull().sum().sum(),
            df.duplicated().sum(),
            df.select_dtypes(include="number").shape[1],
            df.select_dtypes(include="object").shape[1]
        ]
    })
    
    display(ringkasan)

else:
    print("File tidak ditemukan. Periksa lokasi dataset.")
```
