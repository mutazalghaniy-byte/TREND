```python
# Jika library belum terpasang, jalankan di terminal:
# python -m pip install pandas numpy matplotlib seaborn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)

print("Library berhasil diimpor.")
```

```python
# Membaca dataset dari folder yang sama dengan notebook
file_path = Path("klasifikasimhs.csv")

if not file_path.exists():
    raise FileNotFoundError(
        f"File {file_path} tidak ditemukan. "
        "Pastikan CSV berada di folder yang sama dengan notebook."
    )

df = pd.read_csv(file_path)

print("Dataset berhasil dibaca.")
print(f"Jumlah baris  : {df.shape[0]}")
print(f"Jumlah kolom  : {df.shape[1]}")

display(df.head())
```

```python
# Informasi dasar dataset
print("=== INFORMASI DATASET ===")
df.info()

print("\n=== UKURAN DATASET ===")
print(f"Baris  : {df.shape[0]}")
print(f"Kolom  : {df.shape[1]}")

print("\n=== NAMA KOLOM ===")
print(df.columns.tolist())

print("\n=== TIPE DATA ===")
display(df.dtypes.rename("Tipe Data").to_frame())

print("\n=== STATISTIK DESKRIPTIF ===")
display(df.describe(include="all").T)
```

```python
# Ringkasan kualitas data sebelum cleaning
ringkasan_awal = pd.DataFrame({
    "Tipe Data": df.dtypes.astype(str),
    "Data Kosong": df.isna().sum(),
    "Persentase Kosong (%)": (
        df.isna().mean() * 100
    ).round(2),
    "Jumlah Nilai Unik": df.nunique(dropna=True)
})

display(ringkasan_awal)

print("Jumlah baris duplikat:", df.duplicated().sum())

print("\n=== BARIS DENGAN DATA KOSONG ===")
display(df[df.isna().any(axis=1)].head(10))

print("\n=== BARIS DUPLIKAT ===")
display(df[df.duplicated(keep=False)].head(10))
```

```python
# Membuat salinan agar data asli tetap aman
df_clean = df.copy()

# 1. Merapikan nama kolom
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

# 2. Merapikan spasi pada data teks
kolom_teks = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for kolom in kolom_teks:
    df_clean[kolom] = df_clean[kolom].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# 3. Menormalkan label target
kolom_target = "Kelayakan Keringanan UKT"

if kolom_target not in df_clean.columns:
    raise ValueError(f"Kolom target '{kolom_target}' tidak ditemukan.")

# Tampilkan label asli untuk diperiksa
print("Label target sebelum normalisasi:")
display(df_clean[kolom_target].value_counts(dropna=False).to_frame("Jumlah"))

# Asumsi: '1 1' merupakan salah ketik dari '1'.
# Verifikasi asumsi ini dengan sumber data sebelum digunakan.
df_clean[kolom_target] = df_clean[kolom_target].replace({
    "1 1": "1"
})

# 4. Konversi kolom numerik yang seharusnya berupa angka
kolom_numerik = [
    "Tempat Tinggal",
    "Penghasilan Orang Tua",
    "Jumlah Tanggungan Orang Tua",
    "Kendaraan"
]

for kolom in kolom_numerik:
    df_clean[kolom] = pd.to_numeric(
        df_clean[kolom], errors="coerce"
    )

# Konversi target menjadi angka setelah normalisasi
df_clean[kolom_target] = pd.to_numeric(
    df_clean[kolom_target], errors="coerce"
)

# 5. Menghapus baris yang label target-nya tidak valid/kosong.
# Label target tidak diisi menggunakan tebakan statistik.
jumlah_target_tidak_valid = df_clean[kolom_target].isna().sum()

df_clean = df_clean.dropna(subset=[kolom_target]).copy()

print("Baris dengan target tidak valid yang dihapus:",
      jumlah_target_tidak_valid)

# 6. Menghapus duplikat identik
duplikat_dihapus = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("Jumlah baris duplikat yang dihapus:", duplikat_dihapus)

# 7. Menangani nilai kosong pada kolom selain target
for kolom in df_clean.columns:
    if kolom == kolom_target:
        continue

    if df_clean[kolom].isna().any():
        if pd.api.types.is_numeric_dtype(df_clean[kolom]):
            # Median untuk kolom numerik
            df_clean[kolom] = df_clean[kolom].fillna(
                df_clean[kolom].median()
            )
        else:
            # Modus untuk kolom kategorikal
            modus = df_clean[kolom].mode(dropna=True)
            if not modus.empty:
                df_clean[kolom] = df_clean[kolom].fillna(
                    modus.iloc[0]
                )

# 8. Mengatur tipe data akhir
for kolom in kolom_numerik:
    df_clean[kolom] = df_clean[kolom].astype("int64")

# Target merupakan label kelas 0/1
df_clean[kolom_target] = df_clean[kolom_target].astype("int64")

print("Cleaning data selesai.")
print("Ukuran data setelah cleaning:", df_clean.shape)

display(df_clean.head())
```

```python
# Membandingkan kondisi sebelum dan sesudah cleaning
perbandingan = pd.DataFrame({
    "Sebelum Cleaning": [
        len(df),
        df.duplicated().sum(),
        int(df.isna().sum().sum()),
        df.shape[1]
    ],
    "Sesudah Cleaning": [
        len(df_clean),
        df_clean.duplicated().sum(),
        int(df_clean.isna().sum().sum()),
        df_clean.shape[1]
    ]
}, index=[
    "Jumlah Baris",
    "Jumlah Duplikat",
    "Total Data Kosong",
    "Jumlah Kolom"
])

display(perbandingan)

print("\n=== VALIDASI AKHIR ===")
print("Masih ada duplikat:", df_clean.duplicated().any())
print("Masih ada data kosong:", df_clean.isna().any().any())
print("Tipe data akhir:")
display(df_clean.dtypes.rename("Tipe Data").to_frame())

print("\nLabel target setelah cleaning:")
display(df_clean[kolom_target].value_counts().sort_index())
```

```python
# Ringkasan setiap kolom
ringkasan_data = pd.DataFrame({
    "Tipe Data": df_clean.dtypes.astype(str),
    "Jumlah Data": df_clean.count(),
    "Nilai Unik": df_clean.nunique(),
    "Data Kosong": df_clean.isna().sum()
})

display(ringkasan_data)

# Ringkasan statistik kolom numerik
print("=== STATISTIK KOLOM NUMERIK ===")
display(df_clean.describe().T)

# Frekuensi pekerjaan orang tua
print("=== DISTRIBUSI PEKERJAAN ORANG TUA ===")
display(
    df_clean["Pekerjaan Orang Tua"]
    .value_counts()
    .rename_axis("Pekerjaan Orang Tua")
    .reset_index(name="Jumlah")
)

# Distribusi target
print("=== DISTRIBUSI KELAYAKAN UKT ===")
display(
    df_clean[kolom_target]
    .value_counts()
    .sort_index()
    .rename_axis("Kelayakan Keringanan UKT")
    .reset_index(name="Jumlah Mahasiswa")
)
```

```python
# Agregasi berdasarkan pekerjaan orang tua
agregasi_pekerjaan = (
    df_clean.groupby("Pekerjaan Orang Tua")
    .agg(
        Jumlah_Mahasiswa=(kolom_target, "size"),
        Rata_Rata_Penghasilan=("Penghasilan Orang Tua", "mean"),
        Rata_Rata_Tanggungan=("Jumlah Tanggungan Orang Tua", "mean"),
        Total_Penghasilan=("Penghasilan Orang Tua", "sum")
    )
    .round(2)
    .sort_values("Jumlah_Mahasiswa", ascending=False)
)

print("=== AGREGASI BERDASARKAN PEKERJAAN ORANG TUA ===")
display(agregasi_pekerjaan)

# Agregasi berdasarkan label kelayakan (menggunakan variabel kolom_target)
agregasi_ukt = (
    df_clean.groupby(kolom_target) 
    .agg(
        Jumlah_Mahasiswa=(kolom_target, "size"),
        Rata_Rata_Penghasilan=("Penghasilan Orang Tua", "mean"),
        Rata_Rata_Tanggungan=("Jumlah Tanggungan Orang Tua", "mean"),
        Total_Penghasilan=("Penghasilan Orang Tua", "sum")
    )
    .round(2)
    .sort_values("Jumlah_Mahasiswa", ascending=False)
)

print("\n=== AGREGASI BERDASARKAN LABEL KELAYAKAN ===")
display(agregasi_ukt)
```

```python
# Agregasi berdasarkan label kelayakan
agregasi_ukt = (
    df_clean.groupby(kolom_target)
    .agg(
        Jumlah_Mahasiswa=("Penghasilan Orang Tua", "size"),
        Rata_Rata_Penghasilan=("Penghasilan Orang Tua", "mean"),
        Rata_Rata_Tanggungan=("Jumlah Tanggungan Orang Tua", "mean"),
        Rata_Rata_Kendaraan=("Kendaraan", "mean")
    )
    .round(2)
)

print("=== AGREGASI BERDASARKAN KELAYAKAN UKT ===")
display(agregasi_ukt)
```
